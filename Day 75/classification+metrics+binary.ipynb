```python
import pandas as pd 
import numpy as np 
```

```python
df = pd.read_csv("heart.csv")
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df.drop(columns=['target'], axis=1), df['target'], test_size=0.2, random_state=2)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier

clf1 = LogisticRegression()
clf2 = DecisionTreeClassifier()

clf1.fit(X_train, y_train)
clf2.fit(X_train, y_train)

pred1 = clf1.predict(X_test)
pred2 = clf2.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
print(f'Accuracy of {clf1.__class__.__name__}: {accuracy_score(y_test, pred1)}')
print(f'Accuracy of {clf2.__class__.__name__}: {accuracy_score(y_test, pred2)}')
```

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y_test, pred1)*X_test.shape[0]/100
```

```python
print("Logistic Regression Confusion Matrix\n")
pd.DataFrame(confusion_matrix(y_test, pred1),columns=list(range(0,2)))
```

```python

print("Decision Tree Confusion Matrix\n")
pd.DataFrame(confusion_matrix(y_test, pred2),columns=list(range(0,2)))
```

```python
result = pd.DataFrame()
result['Actual Label'] = y_test
result['Logistic Regression Prediction'] = pred1
result['Decision Tree Prediction'] = pred2
```

```python
result.sample(10)
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score
```

```python
print("For Logistic regression Model")
print("-"*50)
cdf = pd.DataFrame(confusion_matrix(y_test,pred1),columns=list(range(0,2)))
print(cdf)
print("-"*50)
print("Precision - ",precision_score(y_test,pred1))
print("Recall - ",recall_score(y_test,pred1))
print("F1 score - ",f1_score(y_test,pred1))
```

```python

print("For DT Model")
print("-"*50)
cdf = pd.DataFrame(confusion_matrix(y_test,pred2),columns=list(range(0,2)))
print(cdf)
print("-"*50)
print("Precision - ",precision_score(y_test,pred2))
print("Recall - ",recall_score(y_test,pred2))
print("F1 score - ",f1_score(y_test,pred2))
```
